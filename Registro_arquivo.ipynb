```python
import time

def openlog(nome_arquivo, modo):
    # Abre o arquivo principal no modo especificado (leitura, escrita, etc.)

    arqEntrada = open(nome_arquivo, modo)

    #obtém hora atual
    now=time.localtime()
    nowFormat=time.strftime('%A %b/%d/%y %I:%M %p', now)

    # Abre o mesmo arquivo novamente no modo 'append' para registrar o log
    arqSaida = open(nome_arquivo, 'a')
    log='{}: nome_arquivo {} aberto.\n'
    arqSaida.write(f'Arquivo {nome_arquivo} aberto no modo {modo}.\n')
    arqSaida.close()

    print('Registro realizado')

    return arqEntrada  # Retorna o arquivo aberto no modo original
```
